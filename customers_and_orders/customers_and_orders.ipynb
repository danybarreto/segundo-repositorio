```python
# open and evaluate orders.csv
datafile = "./sample_data/orders.csv"

if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  valores_unicos = set()
  for row in csvreader:
     valores_unicos.add(row[1])
  freq = Counter(valores_unicos)

print(len(freq))
```

```python
suma_counts = 0
filas_validadas = 0
datafile = "./sample_data/orders.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  for row in csvreader:
        texto = row[4].strip()
        if not texto:
            continue  # salto celdas vacías

        # Ajusta el separador según convenga
        elementos = texto.split(" ")

        suma_counts += len(elementos)
        filas_validadas += 1

if filas_validadas:
    promedio = suma_counts / filas_validadas
    print(f"Promedio de elementos por campo: {promedio:.2f}")
else:
    print("No se encontró ninguna celda válida.")
```

```python
max_count = 0

if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  for row in csvreader:
      cell = row[4].strip()
      if not cell:
            continue

      # Dividir y filtrar posibles cadenas vacías
      elementos = [e for e in cell.split(" ") if e.strip()]
      count = len(elementos)

      if count > max_count:
         max_count = count

print(f"Máximo de elementos en una fila: {max_count}")
```

```python
conteo_exacto = 0
conteo_subcadena = 0

if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  for row in csvreader:
      valor = row[2]

      # Exacto
      if valor == "2021-10-":
         conteo_exacto += 1

      # Subcadena
      if "2021-10-" in valor:
         conteo_subcadena += 1

#print(f"Number of orders placed in October 2021 '2021-10-': {conteo_exacto}")
print(f"Number of orders placed in October 2021: {conteo_subcadena}")
```

```python
id_max = None
valor_max = float("-inf")
datafile = "./sample_data/orders.csv"

if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  headers = next(csvreader)
  for row in csvreader:
      val = float(row[3])
      if val > valor_max and "2021" in row[2]:
          valor_max = val
          id_max = row[0]

print(f"En tabla1 → Orden con valor máximo: {id_max} (valor = {valor_max})")

# 2. Leo los IDs de la tabla B
datafile2 = "./sample_data/customers.csv"

if not exists(datafile2):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile2, 'r') as fl:
  csvreaderb = csv.reader(fl, delimiter=',')
  for row in csvreaderb:
      if row[0] == id_max:
            name_com = row[1] + row[2]
            encontrado = True
            break

if encontrado:
    print(f"Nombre del cliente: ", name_com)
else:
    print(f"ID {id_max} NO se encuentra en tabla2")
```
